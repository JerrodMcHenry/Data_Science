```python
import pandas as pd
```

```python
df = pd.read_csv("Data.csv")
```

Return the average Yearly Amount Spent for each Avatar.

```python
df.groupby("Avatar")["Yearly Amount Spent"].mean()
```

Return how many rows exist per Avatar.

```python
df["Avatar"].value_counts()
```

For each Avatar, return:

Average yearly spend

Maximum yearly spend

```python
df.groupby("Avatar")["Yearly Amount Spent"].agg(
    average_spent = "mean",
    max_spent = "max"
)
```

```python
df.groupby("Avatar")["Yearly Amount Spent"].mean().sort_values(ascending=False)
```
