```python
import pandas as pd
```

```python
df = pd.read_csv('data.csv')
```

```python
df.info()
```

# Step 1 Build a stand alone boolean filter

```python
df["Avatar"] == "Blue"
```

# Step 3: Build filter #2 as a standalone boolean checklist

```python
df["Yearly Amount Spent"] > df["Yearly Amount Spent"].mean()
```

# Step 4: Combine the checklists (and don’t screw it up)

```python
(df["Avatar"] == "Blue") & (df["Yearly Amount Spent"] > df["Yearly Amount Spent"].mean())
```

# Step 5: Apply the combined checklist inside df[ ... ]

```python
df[
(df["Avatar"] == "Blue") & (df["Yearly Amount Spent"] > df["Yearly Amount Spent"].mean())
]
```

```python
df [
(df["Avatar"] == "Blue") & (df["Yearly Amount Spent"] > df["Yearly Amount Spent"].mean())][["\tEmail", "Yearly Amount Spent"]]
```

```python
df [
(df["Avatar"] == "Blue") & (df["Yearly Amount Spent"] > df["Yearly Amount Spent"].mean())][["\tEmail", "Yearly Amount Spent"]].sort_values(by="Yearly Amount Spent", ascending=False)
```

```python
df.loc[:,["\tEmail", "Yearly Amount Spent"]]
```
