```python
import pandas as pd
```

```python
df = pd.read_csv("AI_Impact_on_Jobs_2030.csv")
```

```python
df.info()
```

```python
df.head(10)
```

# Step 2 — loc (label-based selection)

# Example 1 — Select columns by name

```python
df.loc[:,["Average_Salary", "Years_Experience"]]
```

# Example 2 — Filter rows with a condition

```python
df.loc[df["Average_Salary"] == 44363, ["Average_Salary", "Years_Experience"]]
```

# Example 3 — Single row by index label

```python
df.loc[4]
```

# Step 3 — iloc (position-based selection)

# Example 1 — First 5 rows, first 3 columns

```python
df.iloc[:5, :3]
```

# Example 2 — Row 2, column 4

```python
df.iloc[2, 4]
```

# Example 3 — First 10 rows, specific columns

```python
df.iloc[:10, [0,6]]
```

# Step 5 — Missing values (NaN) — guaranteed quiz topic

# Find missing values

```python
df.isna().sum()
```

# Fill missing values (most common)

```python
df["Average_Salary"] = df["Average_Salary"].fillna("Unknown")
```

# Drop rows with missing values (use carefully)

```python
df.dropna()
```

# Or only for one column:

```python
df.dropna(subset=["Average_Salary"])
```

# Step 6 — Fix dirty column names (real-world data)

```python
df.columns = df.columns.str.strip()
```
