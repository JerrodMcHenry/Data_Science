# Step 1 — The simplest groupby

Question:
“What is the average yearly spend per Avatar?”

```python
df.groupby("Avatar")["Yearly Amount Spent"].mean().tail(10)
```

# Step 2 — Turn it into a DataFrame (common quiz move)

If they want a table instead of a Series:

```python
df.groupby("Avatar", as_index=False)["Yearly Amount Spent"].mean().head(10)
```

# Step 3 — Multiple aggregations (this is big-boy Pandas)

Question:
“Show mean and max yearly spend per Avatar.”

```python
df.groupby("Avatar")["Yearly Amount Spent"].agg(["mean", "max"])
```

# Step 4 — Rename columns (clean output)

```python
df.groupby("Avatar")["Yearly Amount Spent"].agg(
    avg_speed = "mean", 
    max_spend = "max"
)
```

# Step 5 — value_counts() (frequency analysis)

```python
df["Avatar"].value_counts().head(5)
```

Equivalent (but more verbose):

```python
df.groupby("Avatar").size().head(5)
```

# Step 6 — Sorting group results

```python
df.groupby("Avatar")["Yearly Amount Spent"].mean().sort_values(ascending = False)
```
