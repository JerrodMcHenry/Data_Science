# Step 1 Build a stand alone boolean filter

```python
df["Avatar"] == "Blue"
```

# Step 3: Build filter #2 as a standalone boolean checklist

```python
df["Yearly Amount Spent"] > df["Yearly Amount Spent"].mean()
```

# Step 4: Combine the checklists (and don’t screw it up)

```python
(df["Avatar"] == "Blue") & (df["Yearly Amount Spent"] > df["Yearly Amount Spent"].mean())
```

# Step 5: Apply the combined checklist inside df[ ... ]

```python
df[
(df["Avatar"] == "Blue") & (df["Yearly Amount Spent"] > df["Yearly Amount Spent"].mean())
]
```
